```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
documents=[
  "I love pizza",
  "Pizza is the best",
  "I love pasta",
  "Pasta is great"
]
```

```python
vectorizer=CountVectorizer()
X=vectorizer.fit_transform(documents)
```

```python
print("Vocabulary: ", vectorizer.get_feature_names_out())#I being a stopword is removed
```

```python
print("\nBoW Matrix\n", X.toarray())
```
