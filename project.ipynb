```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('train.txt',sep=';',header=None,names=['text','emotion'])
```

```python
df
```

```python
#check for null value
df.isnull().sum()
```

```python
#see how many unique emotions
df['emotion'].unique()
```

```python
#assign numbers to emotions
unique_emotions=df['emotion'].unique()
emotion_numbers={}
i=0
for emo in unique_emotions:
  emotion_numbers[emo]=i
  i+=1

df['emotion']=df['emotion'].map(emotion_numbers)
```

```python
df
```

```python
#convert all to lowercse, use lambda function
df['text']=df['text'].apply(lambda x: x.lower())
```

```python
#remove punctuation
import string
def remove_punc(txt):
  return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text']=df['text'].apply(remove_punc)
```

```python
#remove numbers
def remove_num(txt):
  new=''
  for i in txt:
    if not i.isdigit():
      new=new+i
  return new

df['text']=df['text'].apply(remove_num)    
```

```python
#remove emojis by simply checking if its value exists in the ascii code or not
def remove_emojis(txt):
  new=''
  for i in txt:
    if i.isascii():
      new+=i
  return new

df['text']=df['text'].apply(remove_emojis)    
```

```python
#remove stop woords since they add noise
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download("punkt_tab")
```

```python
stop_words=set(stopwords.words('english'))
stop_words
```

```python
#tokenization
def remove(txt):
  words=word_tokenize(txt)
  cleaned=[]
  for i in words:
    if not i in stop_words:
      cleaned.append(i)
  return ' '.join(cleaned)
```

```python
df['text']=df['text'].apply(remove)
```

```python
df.loc[1]['text']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.20, random_state=42)
```

```python
X_train
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
```

```python
bow_vectorizer= CountVectorizer()
X_train_bow= bow_vectorizer.fit_transform(X_train)
X_test_bow=bow_vectorizer.transform(X_test)
#fit_transform() → only for the training set
#transform() → for the test set
```

```python
from sklearn.naive_bayes import MultinomialNB #for text classification
from sklearn.metrics import accuracy_score
```

```python
nb_model=MultinomialNB()
```

```python
nb_model.fit(X_train_bow,y_train)
```

```python
pred_nb=nb_model.predict(X_test_bow)
```

```python
print(accuracy_score(y_test,pred_nb))
```

```python
#tfidf
tfidf_vectorizer= TfidfVectorizer()
X_train_tfidf= tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf=tfidf_vectorizer.transform(X_test)
```

```python
nb_model_tfidf=MultinomialNB()
```

```python
nb_model_tfidf.fit(X_train_tfidf,y_train)
```

```python
pred_nb_tfidf=nb_model.predict(X_test_tfidf)
```

```python
print(accuracy_score(y_test,pred_nb_tfidf))
```
